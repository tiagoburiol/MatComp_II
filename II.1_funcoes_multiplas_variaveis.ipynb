# II.1 Funções de múltiplas variáveis

Uma função de múltiplas variáveis é uma relação matemática que associa a cada elemento de um subconjunto de $\mathbb{R}^n$ um único elemento de $\mathbb{R}^m$, com $n\geq 1$. Quando $m=1$, temos **funções escalares de várias variáveis**, que associam um número real a cada ponto do domínio. Quando $m>1$, temos **funções vetoriais de várias variáveis**, que associam um vetor a cada ponto do domínio. Em particular, quando $n=m$, uma função vetorial $F:D\subseteq\mathbb{R}^n\to\mathbb{R}^n$ pode representar um campo vetorial.

Essas funções ocorrem com frequência em situações práticas. Por exemplo, o volume de um cilindro circular depende do raio da base e da altura. O custo de um produto pode depender do custo dos diferentes materiais utilizados em sua produção, do custo de fabricação e de outros fatores. A temperatura na superfície da Terra pode depender da posição geográfica, da data, da hora, da altitude e de outras variáveis. Uma reação química pode depender das quantidades dos elementos envolvidos, da temperatura e da pressão. Da mesma forma, a velocidade e a direção do vento podem depender do local, da data, da hora e de diversas condições atmosféricas.

Em Ciências e Engenharia, é comum trabalharmos com funções de múltiplas variáveis definidas em domínios bidimensionais ou tridimensionais. Essas funções podem também depender do tempo, que pode ser considerado uma variável adicional. Dependendo da natureza da grandeza representada, a função pode assumir valores escalares ou vetoriais. Por exemplo, a temperatura pode ser representada por uma função escalar, enquanto a velocidade do vento pode ser representada por uma função vetorial, pois possui módulo, direção e sentido.

## Função de duas variáveis reais a uma variável real

Uma função de duas variáveis reais a uma variável real é uma função $f:D\subseteq\mathbb{R}^2\rightarrow\mathbb{R}$ que associa a cada ponto $(x,y)$ do domínio $D$ um único número real $z$. Podemos escrever

$$
z=f(x,y).
$$

Para cada ponto $(x,y)$ do domínio, a função determina um valor $z$, de modo que o gráfico da função é formado pelos pontos $(x,y,z)$ do espaço $\mathbb{R}^3$ que satisfazem a equação $z=f(x,y)$. Assim, o gráfico de uma função de duas variáveis pode ser representado tridimensionalmente como uma superfície no espaço.

Também é possível representar a função de forma bidimensional, por exemplo, por meio de **curvas de nível** ou de um **mapa de cores**, nos quais os valores de $z$ são associados aos diferentes pontos $(x,y)$ do domínio.

Uma função desse tipo é chamada de **função escalar de duas variáveis** e pode ser interpretada como um **campo escalar** definido sobre uma região do plano.

A seguir, vamos considerar alguns casos como exemplos. Para isso vamos utilizar as bibliotecas `numpy` e `matplotlib`, as quais são importadas usando as linhas de código abaixo:

```python
import numpy as np
import matplotlib.pyplot as plt
```

**Exemplo 1:** A distância de um ponto $(x,y)$ do plano cartesiano ao ponto $(2,1)$ é dada pela função de duas variáveis

$$d(x,y)=\sqrt{(x-2)^2+(y-1)^2}.$$

Assim, a distância do ponto $(5,3)$ ao ponto $(2,1)$ é

$$d(5,3)=\sqrt{(5-2)^2+(3-1)^2}=\sqrt{13} \approx 3.60555.$$

Podemos interpretar o valor $d(5,3)=\sqrt{13}$ como a **cota**, ou coordenada $z$, associada ao ponto $(5,3)$. Dessa forma, o ponto $(5,3)$ do plano $xy$ corresponde ao ponto $(5,3,\sqrt{13})$ no espaço $\mathbb{R}^3$, pertencente ao gráfico da função $d$. 

Usando Python podemos definir essa função e calcular valores como é mostrado abaixo:


```python
# define a função
d = lambda x,y: np.sqrt((x-2)**2+(y-1)**2)
```

```python
#Calculando para d(5,3)
d(5,3)
```

De modo geral, para cada ponto $(x,y)$ do domínio, o valor $d(x,y)$ determina uma coordenada $z$. Portanto, o gráfico da função é a superfície do $\mathbb{R}^3$ formada pelos pontos

$$(x,y,d(x,y))=\left(x,y,\sqrt{(x-2)^2+(y-1)^2}\right).$$

Se considerarmos uma região retangular do $\mathbb{R}^2$ como domínio da função, os pontos correspondentes no espaço $\mathbb{R}^3$ formam uma porção dessa superfície.

A seguir, veremos como criar um conjunto de pontos distribuídos em uma grade regular do $\mathbb{R}^2$, calcular os valores da função nesses pontos e representar os resultados graficamente de três modos diferentes.

```python
#cria a malha
x = np.linspace(0, 6)
y = np.linspace(0, 6)
xi, yi = np.meshgrid(x, y)
```

```python
#%matplotlib qt
%matplotlib inline
# cria a figura
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(projection='3d')

# plota superfície
ax.plot_surface(xi, yi, d(xi,yi), alpha=0.5)
ax.plot_wireframe(xi, yi, d(xi,yi), lw=0.5)
ax.view_init(10, 60)
plt.show()
```

```python
# plota mapa de cores 
ax = fig.add_subplot()
plt.contourf(x, y, d(xi,yi), 20)
plt.gca().set_aspect('equal')
plt.colorbar()
plt.show()
```

### Curvas planas em coordenadas polares

Coordenadas polares são um sistema de coordenadas no plano no qual a posição de um ponto é determinada pela distância $r$ até a origem e pelo ângulo $\theta$ que o segmento que liga a origem ao ponto forma com o eixo $x$ positivo. A relação com as coordenadas cartesianas é dada por $x=r\cos\theta$ e $y=r\sin\theta$.

**Exemplo 5:** Considere os pontos dados em coordenadas polares

$$
(r,\theta) = (1,0),\quad
(2,\frac{\pi}{4}),\quad
(3,\frac{\pi}{2}),\quad
(2,\frac{3\pi}{4}),\quad
(1,\pi).
$$

O objetivo é representar esses pontos graficamente utilizando o sistema de coordenadas polares. Para isso, podemos utilizar a biblioteca `matplotlib`, especificando o tipo de projeção `'polar'` na criação dos eixos.

Observe que, no `matplotlib`, as coordenadas são fornecidas na ordem $\theta$ e $r$, isto é, o ângulo é informado primeiro e a distância à origem depois.

A seguir, vamos criar as listas contendo os valores de $r$ e $\theta$ e representar os pontos no plano polar.

```python
r = [1, 2, 3, 2, 1]
theta = [0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi]

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})

ax.scatter(theta, r)

plt.show()
```

Cada ponto do gráfico é determinado pela distância $r$ em relação à origem e pelo ângulo $\theta$. Por exemplo, o ponto $(3,\frac{\pi}{2})$ está a uma distância $3$ da origem e forma um ângulo de $\frac{\pi}{2}$ radianos com o eixo $x$ positivo.

**Exemplo 6:** A **cardioide** é uma curva clássica das coordenadas polares. Uma de suas formas mais simples é dada por

Podemos representar essa curva como é mostrado a seguir:


```python
theta = np.linspace(0, 2*np.pi, 500)
r = 1 + np.cos(theta)

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})

ax.plot(theta, r)

plt.show()
```

**Exemplo 7:** As **rosáceas polares** são curvas que apresentam um padrão de pétalas. Uma família de curvas desse tipo pode ser definida por

O parâmetro $n$ determina a quantidade de pétalas da curva. Para $n$ ímpar, a rosácea possui $n$ pétalas; para $n$ par, possui $2n$ pétalas.

Nesse caso, a curva apresenta $8$ pétalas.

```python
theta = np.linspace(0, 2*np.pi, 1000)
r = np.cos(4*theta)

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})

ax.plot(theta, r)

plt.show()
```

**Exemplo 8:** Representação cartesiana de uma curva polar
Considere a curva polar

$$r=1+\cos (\theta), \quad 0 \leq \theta \leq 2 \pi .$$

Para representar essa curva no sistema cartesiano, primeiro calculamos os valores de $r$ para os diferentes valores de $\theta$. Em seguida, utilizamos as relações entre coordenadas polares e cartesianas,

$$x=r \cos (\theta), \quad y=r \sin (\theta),$$

para obter as coordenadas cartesianas dos pontos da curva.
O gráfico pode então ser construído utilizando os valores de $x$ e $y$:

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 500)

r = 1 + np.cos(theta)

x = r*np.cos(theta)
y = r*np.sin(theta)

plt.plot(x, y)

plt.axis('equal')
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.grid()

plt.show()
```

Aqui, embora a curva seja definida em coordenadas polares pela equação $r=1+\cos(\theta)$, o comando `plt.plot(x, y)` utiliza as coordenadas cartesianas $(x,y)$ para construir o gráfico.

## Função de três variáveis reais a uma variável real

É uma relação $f:R^3 \rightarrow R$ em que o domínio é um subconjunto do espaço tridimensional $\mathbb{R}^3$. Podemos escrever 

$$w = f(x,y,z)$$

assim, os elementos pertencentes a essa função são ponto $(x,y,z,w)$ do $R^4$. Nesse caso não é fácil representar graficamente essa função, cada ponto no espaço assume um valor como imagem. Uma possível representação seria utilizar superfícies de nível, outras abordagens podem envolver planos de corte mapa de cores em um conjunto de pontos e outras técnicas de visualização de dados.

Para cada ponto $(x,y,z)$ do domínio, a função associa um único valor $w$. Assim, seu gráfico é formado pelos pontos $(x,y,z,w)$ do $\mathbb{R}^4$. Como não podemos visualizar diretamente o $\mathbb{R}^4$, utilizaremos diferentes estratégias para representar os valores da função, como cortes, mapas de cores, superfícies de nível e visualizações tridimensionais.

**Exemplo 9:**
Vamos considerar agora a distância de um ponto $(x,y,z)$ do espaço cartesiano ao ponto $(3,2,1)$. Nesse caso, temos uma função de três variáveis

$$d(x,y,z)=\sqrt{(x-3)^2+(y-2)^2+(z-1)^2}.$$

Assim, a distância do ponto $(4,2,3)$ ao ponto $(3,2,1)$ é

$$d(4,2,3)=\sqrt{(4-3)^2+(2-2)^2+(3-1)^2}=\sqrt{5} \approx 2.23607.$$


```python
d = lambda x,y,z: np.sqrt((x-3)**2+(y-2)**2+(z-1)**2)
```

```python
d(4,2,3)
```

```python
z = np.linspace(0, 6, 11)
x = np.linspace(0, 6, 11)
y = np.linspace(0, 6, 11)
xi, yi, zi = np.meshgrid(x, y, z)

d = lambda x,y,z: np.sqrt((x-3)**2+\
                          (y-2)**2+\
                          (z-1)**2)

fig = plt.figure()
ax = plt.axes(projection="3d")

sc = ax.scatter(xi, yi, zi, 
                  c=d(xi,yi,zi), 
                  alpha=0.7, marker='.')
plt.colorbar(sc)
plt.show()
```

```python
w = d(xi,yi,zi)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

for z in [1,3,5]:
    c = ax.contourf(xi[:,:,z], yi[:,:,z], 
                    w[:,:,z], 10, 
                    zdir='z', offset=z,
                   cmap='plasma')
plt.colorbar(c)
ax.set_zlim((0.,7))
plt.show()
```

### Superfícies parametrizadas

**Exemplo 10: (Paraboloide)** 
Gráfico do paraboloide $z(x,y)=x^2+y^2$ discretizado em coordenadas polares $x=r \cos(t)$ e $y = r \text{ sen}(t)$. As mudanças de coordenadas implicam em

$$z=r^2 \cos ^2 t+r^2 \sin ^2 t=r^2 .$$

```python
# usando a função meshgrid
r = 2*np.linspace(0, 1, 11)
t = np.linspace(0, 2*np.pi, 21)

# usando a função meshgrid
rv, tv = np.meshgrid(r, t)

x = rv*np.cos(tv)
y = rv*np.sin(tv)

# define a função
f = lambda x,y: x**2 + y**2
z = f(x,y)

#cria a figura
fig = plt.figure(figsize=(5,5))

# plota
ax = fig.add_subplot(projection='3d')
ax.plot_surface(x, y, z, alpha=0.5)
ax.plot_wireframe(x, y, z, lw=0.5,color='black')
ax.scatter3D(x, y, 0, alpha=0.5,marker='.')
ax.set_zlim(0, 4)
plt.show()
```

Algumas vezes é conveniente representar pontos no espaço $\mathbb{R}^3$ utilizando **coordenadas esféricas**. Nesse sistema de coordenadas, a posição de um ponto é determinada por três parâmetros: a distância $r$ até a origem, o ângulo azimutal $\theta$ e o ângulo polar $\phi$.

A relação entre as coordenadas esféricas $(r,\theta,\phi)$ e as coordenadas cartesianas $(x,y,z)$ é dada por

$$
\begin{aligned}
x &= r\sin(\phi)\cos(\theta),\\
y &= r\sin(\phi)\sin(\theta),\\
z &= r\cos(\phi).
\end{aligned}
$$

em que

- $r\geq 0$ é a distância do ponto à origem;
- $0\leq\theta<2\pi$ é o ângulo azimutal, medido no plano $xy$ a partir do eixo $x$ positivo;
- $0\leq\phi\leq\pi$ é o ângulo polar, medido a partir do eixo $z$ positivo.


**Exemplo 11: (Esfera em coordenadas esféricas)** A distância $\left|P_1 P_2\right|$ entre os pontos $P_1\left(x_1, y_1, z_1\right)$ e $P_2\left(x_2, y_2, z_2\right)$ é

$$
\left|P_1 P_2\right|=\sqrt{\left(x_2-x_1\right)^2+\left(y_2-y_1\right)^2+\left(z_2-z_1\right)^2}
$$

Uma esfera de raio $r$ e centro em $C(h, k, l)$ é definida como o conjunto de todos os pontos $P(x, y, z)$ cuja distância a $C$ é igual a $r$. Assim, a equação de uma esfera com centro $C(h, k, l)$ e raio $r$ é

$$
(x-h)^2+(y-k)^2+(z-l)^2=r^2
$$

Em particular, se o centro é a origem $O$, então a equação da esfera é

$$
x^2+y^2+z^2=r^2
$$

Em coordenadas esféricas, para uma esfera de raio $r$ centrada na
origem, o valor de $r$ é constante, enquanto $\theta$ e $\phi$ variam
em seus respectivos intervalos. Assim, podemos utilizar as coordenadas
esféricas para gerar uma representação paramétrica da superfície da esfera.

```python
%matplotlib inline
```

```python
# Cria as coordenadas esféricas (theta: ângulo azimutal, phi: ângulo polar)
theta = np.linspace(0, 2 * np.pi, 100)   # ângulo ao redor do eixo z
phi = np.linspace(0, np.pi, 100)         # ângulo do polo norte ao sul

# Gera uma malha de coordenadas (grid)
theta, phi = np.meshgrid(theta, phi)

# Raio da esfera
r = 1

# Conversão para coordenadas cartesianas
x = r * np.sin(phi) * np.cos(theta)
y = r * np.sin(phi) * np.sin(theta)
z = r * np.cos(phi)

# Cria a figura e o gráfico 3D
fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(111, projection='3d')

# Plota a superfície
ax.plot_surface(x, y, z, color='skyblue', edgecolor='k', alpha=0.7, lw=0.1)
ax.set_box_aspect([1,1,1])  # eixo com proporções iguais
plt.show()
```

**Exemplo 12: (Cilindro em coordenadas cilíndricas)** As coordenadas cilíndricas são uma extensão das coordenadas polares para o espaço $\mathbb{R}^3$. Nesse sistema, a posição de um ponto é determinada pela distância $r$ até o eixo $z$, pelo ângulo $\theta$ no plano $xy$ e pela coordenada $z$. A relação entre as coordenadas cilíndricas $(r,\theta,z)$ e as coordenadas cartesianas $(x,y,z)$ é dada por

$$
\begin{aligned}
x &= r\cos(\theta),\\
y &= r\sin(\theta),\\
z &= z.
\end{aligned}
$$

em que $r\geq0$ representa a distância do ponto ao eixo $z$, $\theta$ é o ângulo medido no plano $xy$ a partir do eixo $x$ positivo e $z$ representa a altura do ponto em relação ao plano $xy$.

Para representar um cilindro de raio constante, podemos manter $r$ fixo e variar $\theta$ e $z$. A seguir, vamos utilizar essas coordenadas para construir uma representação paramétrica de um cilindro de raio $2$ e altura $4$.

```python
# usando coordenadas cilíndricas

t = np.linspace(0, 2*np.pi, 37)
r = np.full(11, 2)
rv, tv = np.meshgrid(r, t)

x = rv*np.cos(tv)
y = rv*np.sin(tv)

zi = [np.linspace(0,4,11)]
z = np.repeat(zi, len(t), axis=0)

#cria a figura
fig = plt.figure(figsize=(5,5))

# plota
ax = fig.add_subplot(projection='3d')
ax.plot_surface(x, y, z, alpha=0.5)
ax.plot_wireframe(x, y, z, lw=0.5,color='black')

ax.set_zlim(0, 4)
plt.show()
```
