# I.2 SymPy e matrizes

No SymPy, podemos representar vetores e matrizes **simbolicamente** utilizando a classe `sympy.Matrix`, cujos elementos podem, por sua vez, ser representados por **números, símbolos ou até mesmo expressões simbólicas arbitrárias**.

Neste notebook vamos revisitar as operações com vetores e matrizes já estudadas em I.1, mas agora usando o SymPy. A principal vantagem é que os resultados são obtidos de forma **exata** (frações, raízes, expressões simbólicas), sem os erros de arredondamento da aritmética de ponto flutuante usada pelo `numpy`.

Vamos começar revendo vetores, e depois passar para matrizes.

```python
import sympy as sp
```

## Vetores com SymPy

Assim como em I.1, um vetor pode ser representado por suas componentes. Em SymPy, é natural representar um vetor como uma matriz coluna (ou linha) usando `sympy.Matrix`, o que nos permite reaproveitar todos os métodos de matrizes (soma, produto por escalar, etc.) diretamente nos vetores.

**Exemplo 1:** Criando os vetores simbólicos $\vec{u} = [u_1, u_2, u_3]^T$ e $\vec{v} = [v_1, v_2, v_3]^T$.

```python
u = sp.Matrix(sp.symbols('u_1:4'))
v = sp.Matrix(sp.symbols('v_1:4'))

display(u.T, v.T)
```

**Exemplo 2:** Soma, produto por escalar e combinação linear de vetores simbólicos. Os vetores $\vec{u}$ e $\vec{v}$ já foram criados no Exemplo 1. Nesse exemplo vamos incluir as variáveis simbólicas $a$ e $b$ que representam escalares.

```python
a, b = sp.symbols('a b')
display(u + v)        # soma
```

```python
display(a*u)          # produto por escalar
```

```python
display(a*u + b*v)    # combinação linear
```

**Exemplo 3:** Produto escalar (`.dot`), produto vetorial (`.cross`, apenas em $\mathbb R^3$) e norma (`.norm`) de vetores simbólicos.

```python
display(u.dot(v))     # produto escalar
```

```python
display(u.cross(v))   # produto vetorial
```

```python
display(u.norm())     # norma (módulo)
```

**Exemplo 4:** No notebook I.1 usamos `numpy` para calcular o produto escalar, o produto vetorial, a norma e o ângulo entre os vetores $u=(1,2,3)$ e $v=(4,5,6)$, obtendo resultados em ponto flutuante. Vamos repetir esses cálculos com SymPy e comparar.

```python
u = sp.Matrix([1, 2, 3])
v = sp.Matrix([4, 5, 6])
```

```python
u.dot(v)
```

```python
u.cross(v).T
```

```python
u.norm()
```

```python
angulo = sp.acos(u.dot(v)/(u.norm()*v.norm()))
angulo
```

```python
print('ângulo (rad, aproximado) =', angulo.evalf())
```

## Matrizes com SymPy

Vamos agora ver como criar e operar matrizes com SymPy, revisitando as mesmas ideias vistas em I.1 com `numpy`.

**Exemplo 5:** Criando matrizes tipo linha, coluna, quadrada e retangular:

```python
sp.Matrix([[2,1]])
```

```python
sp.Matrix([1, 2])
```

```python
sp.Matrix([[1, 2],[3,4]])
```

Como esse exemplo demonstra, uma única lista gera um **vetor coluna**, enquanto a criação de uma matriz requer uma **lista aninhada** de valores. Diferentemente dos arrays multidimensionais do NumPy, o objeto `sympy.Matrix` do SymPy é destinado apenas a estruturas de até duas dimensões, isto é, vetores e matrizes.

Outra maneira de criar novos objetos `sympy.Matrix` é fornecer como argumentos o número de linhas, o número de colunas e uma função que recebe os índices da linha e da coluna como argumentos e retorna o valor do elemento correspondente:

**Exemplo 6:** Criando uma matriz usando uma função para o termo geral $a_{ij}$:

```python
aij = lambda i, j: 10 * i + j

sp.Matrix(3, 4, aij)
```

**Exemplo 7:** Soma de matrizes e produto por escalar, com elementos simbólicos.

```python
a1,a2,a3,a4 = sp.symbols('a1 a2 a3 a4')
b1,b2,b3,b4 = sp.symbols('b1 b2 b3 b4')

A = sp.Matrix([[a1, a2], [a3, a4]])
B = sp.Matrix([[b1, b2], [b3, b4]])
k = sp.symbols('k')

display(A + B)   # soma
display(k*A)      # produto por escalar
```

**Exemplo 8:** O produto de matrizes, em geral, não comuta: $AB \neq BA$. Vamos verificar isso simbolicamente.

```python
display(A*B)
display(B*A)

sp.simplify(A*B - B*A) == sp.zeros(2,2)
```

Com SymPy é possível criar matrizes com elementos que são **expressões simbólicas**. Por exemplo, uma matriz arbitrária $2 \times 2$ pode ser representada utilizando uma variável simbólica para cada um de seus elementos.

**Exemplo 9:** Criando matrizes com elementos simbólicos e fazendo cálculos com essas matrizes:

```python
a, b, c, d = sp.symbols("a, b, c, d")
M = sp.Matrix([[a, b], [c, d]])

display(M)
```

```python
M*M
```

```python
x = sp.Matrix(sp.symbols("x_1, x_2"))
display(x)
```

```python
M*x
```

**Exemplo 10:** Assim como em I.1, é fácil extrair uma linha de uma matriz, sua diagonal, obter sua transposta ou calcular uma combinação linear de matrizes.

```python
M = sp.Matrix([[a1, a2], [a3, a4]])

display(M.row(0))     # primeira linha
display(M.diagonal())  # diagonal
display(M.T)            # transposta
display(2*M - 3*B)      # combinação linear (M e B definidas nos exemplos anteriores)
```

Algumas funções e métodos para matrizes, disponíveis no SymPy, são mostrados na tabela abaixo:

| Função/Método     | Descrição                                                                                                              |
| :---------------- | :--------------------------------------------------------------------------------------------------------------------- |
| `transpose` / `T` | Calcula a transposta de uma matriz.                                                                                    |
| `adjoint` / `H`   | Calcula a adjunta de uma matriz.                                                                                       |
| `trace`           | Calcula o traço (soma dos elementos da diagonal) de uma matriz.                                                        |
| `det`             | Calcula o determinante de uma matriz.                                                                                  |
| `inv`             | Calcula a inversa de uma matriz.                                                                                       |
| `LUdecomposition` | Calcula a decomposição LU de uma matriz.                                                                               |




**Exemplo 11:** Usando funções do SymPy para operar matrizes.

```python
a,b,c,d = sp.symbols("a,b,c,d")
A = sp.Matrix([[a, b], [c, d]])

display(A)
```

```python
# determinante
A.det()
```

```python
# inversa
A.inv()
```

```python
e,f,g,h,i = sp.symbols("e,f,g,h,i")
M = sp.Matrix([[a,b,c],
               [d,e,f],
               [g,h,i]])
display(M)
```

```python
M.T
```

```python
M.det()
```
