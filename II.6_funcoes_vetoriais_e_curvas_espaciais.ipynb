# II.6 Funções Vetoriais e Curvas Espaciais

Uma função vetorial, ou função com valores vetoriais, é uma função cujo domínio é um conjunto de números reais e cuja imagem é um conjunto de vetores ([Stewart, 2022](https://integrada.minhabiblioteca.com.br/reader/books/9786555584103/pageid/301)).

### Funções vetoriais de uma variável e curvas espaciais

Uma função vetorial de uma variável no espaço é uma relação $\mathbf{r}: I \subset \mathbb{R} \rightarrow \mathbb{R}^3$ que associa a cada número $t$ do domínio $I$ um único vetor $\mathbf{r}(t) \in \mathbb{R}^3$. Se $f(t)$, $g(t)$ e $h(t)$ são as componentes do vetor $\mathbf{r}(t)$, então $f$, $g$ e $h$ são funções com valores reais, chamadas **funções componentes** de $\mathbf{r}$. Denotamos

$$
\mathbf{r}(t)=\langle f(t), g(t), h(t)\rangle=f(t) \mathbf{i}+g(t) \mathbf{j}+h(t) \mathbf{k}.
$$

As curvas espaciais e as funções vetoriais contínuas estão intimamente relacionadas. Suponha que $f$, $g$ e $h$ sejam funções reais contínuas em um intervalo $I$. Então, o conjunto $C$ de todos os pontos $(x, y, z)$ no espaço tais que

$$
x=f(t) \quad y=g(t) \quad z=h(t)
$$

com $t$ variando no intervalo $I$, é chamado **curva espacial**. As equações acima são denominadas **equações paramétricas** de $C$ e $t$ é conhecido como **parâmetro**. Assim, o gráfico de uma função vetorial é uma curva paramétrica no espaço, cujos pontos dependem do parâmetro $t$.

### Plotando funções vetoriais e curvas paramétricas

```python
import numpy as np
import matplotlib.pyplot as plt
```

**Exemplo 1:** Nesse exemplo vamos representar graficamente a função vetorial

$$\mathbf{r}(t) = \left\langle t^2 \operatorname{sen} t,\ t\cos t,\ \sqrt{t} \right\rangle$$

com $0 \leq t \leq 8 \pi$.

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

t = np.linspace(0, 8*np.pi, 400)
x = t**2 * np.sin(t)
y = t*np.cos(t)
z = np.sqrt(t)

ax.plot(x, y, z)
plt.show()
```

**Exemplo 4:** ([Stewart, 2022, p. 824](https://integrada.minhabiblioteca.com.br/reader/books/9786555584103/pageid/303)) Determine uma equação vetorial que represente a curva obtida pela intersecção do cilindro $x^2+y^2=1$ com o plano $y+z=2$.

Podemos escrever as equações paramétricas para $C$ como

$$
x=\cos t \quad y=\operatorname{sen} t \quad z=2-\operatorname{sen} t \quad 0 \leq t \leq 2 \pi
$$

então, a equação vetorial correspondente é 

$$\mathbf{r}(t)=\cos t \mathbf{i}+\operatorname{sen} t \mathbf{j}+(2-\operatorname{sen} t) \mathbf{k} \quad 0 \leq t \leq 2 \pi$$

```python
#cria a figura
fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(projection='3d')

# cilindro - coordenadas cilindricas
t,z = np.meshgrid(np.linspace(0,2*np.pi,36), np.linspace(0,5,20))
x = np.cos(t)
y = np.sin(t)
ax.plot_wireframe(x, y, z, lw=0.2, color='k')

# plano
x,y = np.meshgrid(np.linspace(-1,1,20), np.linspace(-1,1,20))
z = 2-y
ax.plot_wireframe(x, y, z, lw=0.2, color='k')

t = np.linspace(0,2*np.pi,100)
x = np.cos(t)
y = np.sin(t)
z = 2-np.sin(t)
ax.plot(x, y, z, lw=1, color='r')
plt.show()
```

**Exemplo 5:** ([Stewart, 2022, p. 825](https://integrada.minhabiblioteca.com.br/reader/books/9786555584103/pageid/303)) Determine as equações paramétricas da curva formada pela intersecção do paraboloide $4 y=x^2+z^2$ com o plano $y=x$.

*SOLUÇÃO:* Como qualquer ponto da curva $C$, que está na intersecção das superfícies, satisfaz as equações das duas superfícies, podemos substituir $y=x$ na equação do paraboloide, o que nos fornece $4 x=x^2+z^2$. Completando o quadrado em $x$, obtemos $(x-2)^2+z^2=4$, de modo que $C$ deve estar contida no cilindro circular $(x-2)^2+z^2=4$ e a projeção de $C$ no plano $x z$ será a circunferência definida por $(x-2)^2+z^2=4, y=0$, que tem centro em $(2,0,0)$ e raio 2. Assim, podemos escrever $x=2+2 \cos t$, $z=2 \operatorname{sen} t, 0 \leq t \leq 2 \pi$, e, uma vez que $y=x$, as equações paramétricas de $C$ são

$$
x=2+2 \cos t, \quad y=2+2 \cos t, \quad z=2 \operatorname{sen} t, \quad 0 \leq t \leq 2 \pi
$$

```python
#cria a figura
fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(projection='3d')

# paraboloide
t = np.linspace(0, 2*np.pi, 37)
r = np.linspace(0, 5)
rv, tv = np.meshgrid(r, t)

x = rv*np.cos(tv)
z = rv*np.sin(tv)
y = x**2/4 + z**2/4

#ax.plot_surface(x, y, z, alpha=0.5)
ax.plot_wireframe(x, y, z, lw=0.2, color='k')

# plano
x, z = np.meshgrid(np.linspace(0,5,21), np.linspace(-5,5,11))
y = x.copy()

#ax.plot_surface(x, y, z, alpha=0.5)
ax.plot_wireframe(x, y, z, lw=0.2, color='k')

# curva
t = np.linspace(0, 2*np.pi, 100)

x = 2+2*np.cos(t)
y = 2+2*np.cos(t)
z = 2*np.sin(t)

ax.plot(x, y, z, lw=1, color='r')

ax.set_box_aspect([1,1,1])
plt.show()
```

### Comprimento de uma curva

Se $C$ é a curva representada por $\mathbf{r}(t)$, com derivada contínua $\mathbf{r}^{\prime}(t)$ em $a \leq t \leq b$, e se $C$ é percorrida exatamente uma vez quando $t$ aumenta de $a$ para $b$, então o comprimento $L$ de $C$ é

$$
L=\int_a^b \left|\mathbf{r}^{\prime}(t)\right| d t = \int_a^b \sqrt{\mathbf{r}^{\prime}(t) \cdot \mathbf{r}^{\prime}(t)}\, d t.
$$

**Exemplo 9:** Vamos calcular o comprimento do arco da hélice circular $\mathbf{r}(t)=\langle 2 \cos t, 2 \operatorname{sen} t, 6 t \rangle$ do ponto $(2,0,0)$ até o ponto $(2,0,24 \pi)$. Primeiro, determinamos os valores do parâmetro $t$ correspondentes a esses pontos:

```python
import sympy as sp
from sympy import sin, cos, pi
```

```python
t = sp.symbols('t')
```

```python
r = sp.Matrix([2*cos(t),2*sin(t),6*t])
sp.solve(sp.Eq(r, sp.Matrix([2,0,0])))
```

```python
sp.solve(sp.Eq(r, sp.Matrix([2,0,24*pi])))
```

Em seguida, calculamos $\left|\mathbf{r}^{\prime}(t)\right|$ e simplificamos:

```python
norm_dr = sp.sqrt(r.diff(t).dot(r.diff(t)))
display(norm_dr)
```

```python
norm_dr = sp.trigsimp(norm_dr)
norm_dr
```

Por fim, integramos de $t=0$ até $t=4\pi$:

```python
sp.integrate(norm_dr, (t,0,4*pi))
```

### Campos vetoriais

([Stewart, 2022](https://integrada.minhabiblioteca.com.br/reader/books/9786555584103/pageid/518)) Seja $D$ um conjunto em $\mathbb{R}^2$ (uma região plana). Um campo vetorial em $\mathbb{R}^2$ é uma função $\mathbf{F}$ que associa a cada ponto $(x, y)$ em $D$ um vetor bidimensional $\mathbf{F}(x, y)$.

Uma vez que $\mathbf{F}(x, y)$ é um vetor bidimensional, podemos escrevê-lo em termos de suas funções componentes $P$ e $Q$ da seguinte forma:

$$
\mathbf{F}(x, y)=P(x, y) \mathbf{i}+Q(x, y) \mathbf{j}=\langle P(x, y), Q(x, y)\rangle
$$

ou, de forma mais compacta,

$$
\mathbf{F}=P \mathbf{i}+Q \mathbf{j}
$$


Observe que $P$ e $Q$ são funções escalares de duas variáveis e são chamadas, algumas vezes, campos escalares, para distingui-los dos campos vetoriais.

**Exemplo 10:**
Nesse exemplo vamos representar graficamente o campo de vetores dado pela função

$$\mathbf{F}(x,y)=\left(\frac{-y^2}{\sqrt{x^4+y^4}},\frac{x}{\sqrt{x^2+y^2}} \right)$$

usando uma malha de $10 \times 10$ pontos, com $-5 \leq x \leq 5$ e $-5 \leq y \leq 5$. Note que, com 10 pontos em cada direção, a malha não contém a origem, onde $\mathbf{F}$ não está definido. Primeiro, usamos a função `quiver`, que desenha uma seta em cada ponto da malha:

```python
x = np.linspace(-5,5,10)
y = np.linspace(-5,5,10)
xi,yi = np.meshgrid(x,y)

u = -yi**2/np.sqrt(xi**4 + yi**4)
v = xi/np.sqrt(xi**2 + yi**2)

fig = plt.figure(figsize=(3,3))
plt.quiver(xi, yi, u, v, scale_units='xy', angles='xy', width=0.004)
plt.show()
```

Outra forma de visualizar o campo é por meio de linhas de fluxo, que são curvas tangentes aos vetores do campo em cada ponto. Para isso usamos a função `streamplot`:

```python
fig = plt.figure(figsize=(3,3))
plt.streamplot(xi,yi,u,v)
plt.show()
```

Seja $E$ um subconjunto de $\mathbb{R}^3$. Um campo vetorial em $\mathbb{R}^3$ é uma função $\mathbf{F}$ que associa a cada ponto $(x, y, z)$ em $E$ um vetor tridimensional $\mathbf{F}(x, y, z)$.

Podemos escrever um campo vetorial $\mathbf{F}$ em $\mathbb{R}^3$ em termos das funções componentes $P, Q$ e $R$ como

$$
\mathbf{F}(x, y, z)=P(x, y, z) \mathbf{i}+Q(x, y, z) \mathbf{j}+R(x, y, z) \mathbf{k}
$$

**Exemplo 11:** Nesse exemplo vamos representar graficamente o campo de vetores dado pela função

$$\mathbf{F}(x,y,z)=\left( \operatorname{sen} x \cos (yz),\ -yz,\ \cos x \operatorname{sen} z \right)$$

usando uma malha de $8 \times 8 \times 8$ pontos, com $-1 \leq x \leq 1$, $-1 \leq y \leq 1$ e $-1 \leq z \leq 1$.

```python
fig = plt.figure(figsize=(5,5))
ax = fig.add_subplot(projection='3d')

x,y,z = np.meshgrid(np.linspace(-1, 1, 8),
                    np.linspace(-1, 1, 8),
                    np.linspace(-1, 1, 8))

u = np.sin(x) * np.cos(y*z)
v = -y*z
w = np.cos(x) * np.sin(z)

ax.quiver(x, y, z, u, v, w,
          length=0.2, linewidths=0.7,
          color='black')

plt.show()
```

### Campo gradiente
Se $f$ é uma função escalar de duas variáveis, seu gradiente $\nabla f$ (ou $\operatorname{grad} f$) é definido por

$$
\nabla f(x, y)=f_x(x, y) \mathbf{i}+f_y(x, y) \mathbf{j}
$$

Portanto, $\nabla f$ é realmente um campo vetorial em $\mathbb{R}^2$ e é denominado campo vetorial gradiente. Da mesma forma, se $f$ for uma função escalar de três variáveis, seu gradiente é um campo vetorial em $\mathbb{R}^3$ dado por

$$
\nabla f(x, y, z)=f_x(x, y, z) \mathbf{i}+f_y(x, y, z) \mathbf{j}+f_z(x, y, z) \mathbf{k}
$$

**Exemplo 12:** Vamos representar os gráficos de três funções de duas variáveis (campos escalares) e, em seguida, os seus campos gradiente (campos vetoriais) juntamente com as curvas de nível. As funções são

$$f_1(x,y) = x^2+y^2, \qquad f_2(x,y) = x^2-y^2, \qquad f_3(x,y) = x^4+y^4-4xy+1.$$

```python
f1 = lambda x,y: x**2+y**2
f2 = lambda x,y: x**2-y**2
f3 = lambda x,y: x**4+y**4-4*x*y+1
```

```python
xi, yi = np.meshgrid(np.linspace(-5,5,201),
                     np.linspace(-5,5,201))

#criamos a figura e o subplot 3d
fig = plt.figure(figsize=(13,4))

ax = fig.add_subplot(131, projection='3d')
ax.plot_surface(xi, yi, f1(xi,yi))

ax = fig.add_subplot(132, projection='3d')
ax.plot_surface(xi, yi, f2(xi,yi))

ax = fig.add_subplot(133, projection='3d')
ax.plot_surface(xi, yi, f3(xi,yi))

plt.show()
```

Para obter o campo gradiente numericamente, usamos a função `gradient` do NumPy. Ela retorna as derivadas parciais na ordem dos eixos do array: como as linhas da matriz correspondem a $y$ e as colunas a $x$, o primeiro array retornado é $f_y$ e o segundo é $f_x$. Passamos também os vetores `x` e `y` para que as derivadas sejam calculadas com o espaçamento correto da malha.

```python
x = np.linspace(-5,5,21)
y = np.linspace(-5,5,21)
xi, yi = np.meshgrid(x, y)

fig = plt.figure(figsize=(13,4))

ax = fig.add_subplot(131)
v, u = np.gradient(f1(xi,yi), y, x)
ax.quiver(xi, yi, u, v, color='b')
ax.contour(xi, yi, f1(xi,yi), 10, colors="r", linewidths=0.4)
ax.set_title(r'$f_1(x,y)=x^2+y^2$')

ax = fig.add_subplot(132)
v, u = np.gradient(f2(xi,yi), y, x)
ax.quiver(xi, yi, u, v, color='b')
ax.contour(xi, yi, f2(xi,yi), 10, colors="r", linewidths=0.4)
ax.set_title(r'$f_2(x,y)=x^2-y^2$')

ax = fig.add_subplot(133)
v, u = np.gradient(f3(xi,yi), y, x)
ax.quiver(xi, yi, u, v, color='b')
xf, yf = np.meshgrid(np.linspace(-5,5), np.linspace(-5,5))
niveis = [-0.9, 0, 1, 2, 10, 50, 100, 200, 400, 800]
ax.contour(xf, yf, f3(xf,yf), niveis, colors="r", linewidths=0.4)
ax.set_title(r'$f_3(x,y)=x^4+y^4-4xy+1$')

plt.show()
```

Observe que, em cada ponto, o vetor gradiente é perpendicular à curva de nível que passa por aquele ponto e aponta no sentido de crescimento da função.
